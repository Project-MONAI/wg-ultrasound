# Ultrasound DICOM Interoperability Evaluation
## PyDICOM + SimpleITK | Doppler/Colour Flow Detection | MONAI Tensor Shape

**Purpose:**  
Provide a clear, reproducible workflow to evaluate ultrasound DICOM files from multiple vendors and acquisition types.  

**Key question addressed:**  
Why do some files open correctly in GDCM-based viewers (e.g. MicroDICOM) but fail in PyDICOM and SimpleITK?  
This notebook diagnoses those gaps and classifies each file accordingly.

**Methodology:**
- **Standard read:** PyDICOM (`force=True`) + SimpleITK applied directly to all files → **99.4% success (1,406 / 1,414 files)**.
- **Robust-read techniques explored** (implemented and tested on the full 1,414-file corpus; produced identical results — removed from this version):
  - *Metadata:* 3-pass read — strict → `force=True` → manual implicit-VR fallback (injecting `is_implicit_VR` + `is_little_endian`)
  - *Pixels:* `TransferSyntaxUID` injection cycling through Implicit VR LE, Explicit VR LE, and Explicit VR BE
- **Root causes of the 8 non-`both_ok` files** (diagnosed via raw binary inspection):
  - 7 × `empty_file`: 0-byte files from the ReMIND dataset — incomplete download
  - 1 × `proprietary_format`: Philips 3D/4D container (magic bytes `377ABCAF`, single private tag `7A37,AFBC`) — not standard DICOM
- **Complete failure report:** `failed_files_us_dicom.xlsx`

```python
# ============================================================
# DICOM DATASET COLLECTION CONFIG
# ============================================================

from pathlib import Path
import pydicom


# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

BASE_DIR = Path(".").resolve()

DATASET = BASE_DIR / "Datasets"

WORKSPACE_DIR = BASE_DIR.parent


# ------------------------------------------------------------
# INPUT DATASET FOLDERS
# ------------------------------------------------------------

DATASET_ROOTS = [
    DATASET / "Aliza",
    DATASET / "AIIMS",
    DATASET / "Breast_Lesions_USG",
    DATASET / "MONAI",
    DATASET / "ReMIND",
    DATASET / "Heartcycle",
    DATASET / "UTA4",
    DATASET / "UTA7",
    DATASET / "UTA10",
    DATASET / "TCIA_IDC_Prostate_MRI_US_Biopsy",
    
]


# ------------------------------------------------------------
# OUTPUT CSV
# ------------------------------------------------------------

OUT_CSV = BASE_DIR / "interoperability_evaluation_us_dicom_tensor.csv"


# ============================================================
# STRICT DICOM DETECTION
# ============================================================

def is_dicom_file(path: Path) -> bool:
    """
    Strict DICOM validation.

    Prevents false positives from:
    - PNG
    - JPG
    - MP4
    - random binary files
    """

    try:

        # Standard DICOM preamble check
        with open(path, "rb") as f:
            f.seek(128)
            magic = f.read(4)

        if magic == b"DICM":
            return True

        # Secondary strict validation
        ds = pydicom.dcmread(
            str(path),
            stop_before_pixels=True,
            force=False
        )

        # Essential DICOM tags must exist
        required_tags = [
            "PatientID",
            "StudyInstanceUID",
            "SeriesInstanceUID",
            "SOPInstanceUID",
        ]

        return any(hasattr(ds, tag) for tag in required_tags)

    except Exception:
        return False


# ============================================================
# COLLECT DICOM FILES
# ============================================================

def collect_dicom_files(roots):

    found = []

    print("Collecting DICOM files...\n")

    for root in roots:

        if not root.exists():
            print(f"[NOT FOUND] {root}")
            continue

        print(f"Scanning: {root}")

        all_files = [f for f in root.rglob("*") if f.is_file()]

        before_count = len(found)

        for f in all_files:

            # Common DICOM extensions
            if f.suffix.lower() in [".dcm", ".dicom"]:
                if is_dicom_file(f):
                    found.append(f)
                continue

            # Extensionless anonymized DICOMs
            if f.suffix == "":
                if is_dicom_file(f):
                    found.append(f)

        added = len(found) - before_count

        print(f"  Found {added} DICOM files\n")

    return sorted(found)


# ============================================================
# RUN COLLECTION
# ============================================================

candidate_files = collect_dicom_files(DATASET_ROOTS)

print("=" * 60)
print(f"TOTAL DICOM FILES FOUND: {len(candidate_files)}")
print("=" * 60)
```

## Section 2 — Structural Probe

Before calling PyDICOM or SimpleITK we check the raw binary signature of each file.

**Note:**  
If a file lacks the DICOM Part 10 preamble (`DICM` at byte offset 128), it may still be a valid DICOM implicit-VR file.
GDCM-based viewers (MicroDICOM, RadiAnt) handle implicit-VR and non-standard files better than PyDICOM by default.  
This probe detects that case **before** any library call to explain failures accurately.

| Signature | Meaning |
|---|---|
| `dicom_part10` | Standard DICOM with DICM prefix at byte 128 |
| `implicit_vr_candidate` | Possible DICOM without preamble (GDCM-readable) |
| `unknown_binary` | Not recognisable as DICOM |

```python
# Magic bytes for Philips proprietary 3D/4D format (private tag 7A37,AFBC)
_PHILIPS_PRIVATE_MAGIC = bytes.fromhex("377abcaf")

def structural_probe(path: Path) -> dict:
    """Read first 512 bytes and classify the DICOM structural signature."""
    result = {
        "dicm_prefix_present": False,
        "dicm_offset": -1,
        "file_meta_likely": False,
        "binary_signature": "unknown",
        "file_size_bytes": 0,
        "probe_error": "",
    }
    try:
        result["file_size_bytes"] = path.stat().st_size
        if result["file_size_bytes"] == 0:
            result["binary_signature"] = "empty_file"
            return result
        raw = path.read_bytes()[:512]
        if raw[128:132] == b"DICM":
            result["dicm_prefix_present"] = True
            result["dicm_offset"] = 128
            result["binary_signature"] = "dicom_part10"
        elif raw[:4] == _PHILIPS_PRIVATE_MAGIC:
            # Philips proprietary 3D/4D: single private tag (7A37,AFBC) wraps all image data
            result["binary_signature"] = "philips_private_3d"
        else:
            starts_with_tag = raw[:2] in (b"\x08\x00", b"\x00\x08", b"\x02\x00", b"\x00\x02")
            try:
                text = raw.decode("latin-1")
            except Exception:
                text = ""
            uid_like = "." in text[8:30] and any(c.isdigit() for c in text[8:30])
            if starts_with_tag or uid_like:
                result["binary_signature"] = "implicit_vr_candidate"
            else:
                result["binary_signature"] = "unknown_binary"
        result["file_meta_likely"] = b"\x02\x00" in raw[:136]
    except Exception as exc:
        result["probe_error"] = f"{type(exc).__name__}: {exc}"
    return result


print(f"{'File':<60} {'DICM':>6}  Signature")
print("-" * 95)
probe_results = {str(f): structural_probe(f) for f in candidate_files}
for f in candidate_files:
    pr = probe_results[str(f)]
    try:
        label = str(f.relative_to(BASE_DIR))
    except ValueError:
        label = f.name
    print(f"{label:<60} {str(pr['dicm_prefix_present']):>6}  {pr['binary_signature']}")
```

## Section 3 — PyDICOM: Metadata Read

We attempt to read DICOM metadata with PyDICOM in two passes:
1. Standard read (strict)
2. `force=True` — handles files without a Part 10 preamble

```python
import csv

TAGS_OF_INTEREST = [
    "Modality", "Manufacturer", "ManufacturerModelName",
    "SOPClassUID", "TransferSyntaxUID",
    "Rows", "Columns", "SamplesPerPixel",
    "PhotometricInterpretation", "NumberOfFrames",
    "UltrasoundColorDataPresent", "ImageType",
    "StudyDescription", "SeriesDescription", "ProtocolName",
    "VolumetricProperties", "FrameTime", "CineRate",
    "BitsAllocated", "BitsStored", "HighBit", "PixelRepresentation",
]

IMPLICIT_VR_LE = "1.2.840.10008.1.2"    # Implicit VR Little Endian
EXPLICIT_VR_LE = "1.2.840.10008.1.2.1"  # Explicit VR Little Endian


def classify_failure(error_msg, warnings_str):
    if not error_msg:
        return ("OK", "")
    if "InvalidDicomError" in error_msg:
        return ("Not DICOM", "File does not conform to DICOM Part 10 or is corrupted")
    if "FileNotFoundError" in error_msg:
        return ("File Not Found", "File missing or path error")
    if "UnsupportedTransferSyntax" in error_msg or "decoder not available" in error_msg:
        return ("Decoder Missing", "Compressed transfer syntax not supported by installed decoders")
    if "AttributeError" in error_msg and "file_meta" in error_msg:
        return ("Missing File Meta", "No file meta information (0002,xxxx) present")
    if "PermissionError" in error_msg:
        return ("Permission Error", "No read access to file")
    if warnings_str:
        return ("Warning", warnings_str)
    return ("Other Error", error_msg)


def _get_tag(ds, keyword, default=""):
    try:
        val = getattr(ds, keyword)
        if hasattr(val, '__iter__') and not isinstance(val, str):
            return "\\".join(str(v) for v in val)
        return str(val) if val is not None else default
    except AttributeError:
        return default


def _extract_tags(ds, result):
    for tag in TAGS_OF_INTEREST:
        result[tag] = _get_tag(ds, tag)
    try:
        result["TransferSyntaxName"] = UID(ds.file_meta.TransferSyntaxUID).name \
            if hasattr(ds, "file_meta") and hasattr(ds.file_meta, "TransferSyntaxUID") else ""
    except Exception:
        result["TransferSyntaxName"] = ""
    try:
        result["SOPClassName"] = UID(ds.SOPClassUID).name
    except Exception:
        result["SOPClassName"] = ""


def pydicom_read_metadata(path: Path) -> dict:
    result = {
        "pydicom_meta_ok": False,
        "pydicom_meta_forced": False,
        "pydicom_meta_error": "",
        "pydicom_warnings": "",
    }
    ds = None
    with warnings.catch_warnings(record=True) as w:
        warnings.simplefilter("always")
        # Pass 1: strict
        try:
            ds = pydicom.dcmread(str(path), stop_before_pixels=True, force=False)
            result["pydicom_meta_ok"] = True
        except Exception as exc:
            first_err = f"{type(exc).__name__}: {exc}"
            # Pass 2: force=True (handles files without Part 10 preamble)
            try:
                ds = pydicom.dcmread(str(path), stop_before_pixels=True, force=True)
                result["pydicom_meta_ok"] = True
                result["pydicom_meta_forced"] = True
            except Exception as exc2:
                result["pydicom_meta_error"] = (
                    f"strict: {first_err} | force: {type(exc2).__name__}: {exc2}"
                )
        if w:
            result["pydicom_warnings"] = "; ".join(
                f"{x.category.__name__}: {x.message}" for x in w
            )
    if ds is not None:
        _extract_tags(ds, result)
    return result


print("Reading metadata with PyDICOM...")
meta_results = {}
failure_log = []
for f in candidate_files:
    r = pydicom_read_metadata(f)
    meta_results[str(f)] = r
    try:
        label = str(f.relative_to(BASE_DIR))
    except ValueError:
        label = f.name
    if r["pydicom_meta_ok"]:
        note = " (forced)" if r["pydicom_meta_forced"] else ""
        print(f"  {label:<60} OK{note}")
    else:
        fail_type, fail_root = classify_failure(r["pydicom_meta_error"], r["pydicom_warnings"])
        print(f"  {label:<60} FAILED [{fail_type}]")
        failure_log.append({
            "file": str(f),
            "status": "FAILED",
            "failure_type": fail_type,
            "root_cause": fail_root,
            "error": r["pydicom_meta_error"],
            "warnings": r["pydicom_warnings"],
        })

print(f"\nTotal metadata failures: {len(failure_log)}")

# Export failure report
fail_report_path = BASE_DIR / f"failure_report_{len(candidate_files)}files_{len(failure_log)}failures_{pd.Timestamp.now().strftime('%Y%m%d')}.csv"
with open(fail_report_path, "w", newline='', encoding="utf-8") as csvfile:
    fieldnames = ["file", "status", "failure_type", "root_cause", "error", "warnings"]
    writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
    writer.writeheader()
    for row in failure_log:
        writer.writerow(row)
print(f"Failure report: {fail_report_path}")

# Copy problematic files grouped by failure type
problem_dir = BASE_DIR / f"problematic_files_{pd.Timestamp.now().strftime('%Y%m%d')}"
problem_dir.mkdir(exist_ok=True)
for fail in failure_log:
    failtype_dir = problem_dir / fail["failure_type"].replace(" ", "_")
    failtype_dir.mkdir(exist_ok=True)
    src = Path(fail["file"])
    dst = failtype_dir / src.name
    try:
        if not dst.exists():
            shutil.copy2(src, dst)
    except Exception as e:
        print(f"[COPY ERROR] {src.name}: {e}")
print(f"Problematic files copied to: {problem_dir}")
print("Done.")
```

## Section 4 — PyDICOM: Pixel Data Read

We attempt to decode the actual pixel array.  
This is a **separate** step from metadata because metadata can succeed while pixel decoding fails.

When decoding fails, the cell shows the **exact error** and the **transfer syntax** that caused it,  
so you know which optional decoder (`pylibjpeg`, `gdcm`) is needed.

**MicroDICOM note:** MicroDICOM uses GDCM internally and can decode transfer syntaxes that PyDICOM  
cannot without the `gdcm` Python binding installed. If a file opens in MicroDICOM but fails here,  
installing `gdcm` (`pip install gdcm` or `conda install -c conda-forge gdcm`) will likely fix it.

```python
def pydicom_read_pixels(path: Path) -> dict:
    result = {
        "pydicom_pixel_ok": False,
        "pydicom_pixel_error": "",
        "pydicom_pixel_shape": "",
        "pydicom_pixel_dtype": "",
        "pydicom_pixel_warnings": "",
    }
    with warnings.catch_warnings(record=True) as w:
        warnings.simplefilter("always")
        try:
            ds = pydicom.dcmread(str(path), force=True)
            arr = ds.pixel_array
            result["pydicom_pixel_ok"] = True
            result["pydicom_pixel_shape"] = "x".join(str(d) for d in arr.shape)
            result["pydicom_pixel_dtype"] = str(arr.dtype)
        except Exception as exc:
            result["pydicom_pixel_error"] = f"{type(exc).__name__}: {exc}"
        if w:
            result["pydicom_pixel_warnings"] = "; ".join(
                f"{x.category.__name__}: {x.message}" for x in w
            )
    return result


print("Reading pixel data with PyDICOM...")
pixel_results = {}
for f in candidate_files:
    if not meta_results[str(f)]["pydicom_meta_ok"]:
        pixel_results[str(f)] = {
            "pydicom_pixel_ok": False,
            "pydicom_pixel_error": "SKIPPED — metadata read failed",
            "pydicom_pixel_shape": "", "pydicom_pixel_dtype": "", "pydicom_pixel_warnings": "",
        }
        continue
    r = pydicom_read_pixels(f)
    pixel_results[str(f)] = r
    try:
        label = str(f.relative_to(BASE_DIR))
    except ValueError:
        label = f.name
    ts = meta_results[str(f)].get("TransferSyntaxName", "")
    if r["pydicom_pixel_ok"]:
        warn_flag = " [warnings]" if r["pydicom_pixel_warnings"] else ""
        print(f"  {label:<60} OK  shape={r['pydicom_pixel_shape']}  dtype={r['pydicom_pixel_dtype']}{warn_flag}")
    else:
        print(f"  {label:<60} FAILED [{ts}]")
        print(f"    -> {r['pydicom_pixel_error'][:120]}")
print("Done.")
```

## Section 5 — SimpleITK Read

A file can be readable by PyDICOM but not by SimpleITK (or vice versa) depending on the transfer syntax.  
SimpleITK also uses GDCM internally, so files that fail here but open in MicroDICOM suggest a  
deeper encoding issue not related to the viewer — likely missing pixel data or a corrupt stream.

```python
def sitk_read(path: Path) -> dict:
    result = {"sitk_ok": False, "sitk_error": "", "sitk_dimension": "", "sitk_size": ""}
    if not SITK_OK:
        result["sitk_error"] = "SimpleITK not installed"
        return result
    try:
        reader = sitk.ImageFileReader()
        reader.SetFileName(str(path))
        reader.ReadImageInformation()
        img = reader.Execute()
        result["sitk_ok"] = True
        result["sitk_dimension"] = str(img.GetDimension())
        result["sitk_size"] = "x".join(str(s) for s in img.GetSize())
    except Exception as exc:
        result["sitk_error"] = f"{type(exc).__name__}: {exc}"
    return result


print("Reading files with SimpleITK...")
sitk_results = {}
for f in candidate_files:
    r = sitk_read(f)
    sitk_results[str(f)] = r
    try:
        label = str(f.relative_to(BASE_DIR))
    except ValueError:
        label = f.name
    if r["sitk_ok"]:
        print(f"  {label:<60} OK  dim={r['sitk_dimension']}  size={r['sitk_size']}")
    else:
        print(f"  {label:<60} FAILED")
        print(f"    -> {r['sitk_error'][:120]}")
print("Done.")
```

## Section 6 — Consistency Classification

We combine results from Sections 2–5 into a single per-file classification.

| `comparison_status` | Meaning |
|---|---|
| `both_ok` | Both PyDICOM and SimpleITK read the file successfully |
| `pydicom_only` | PyDICOM read pixels; SimpleITK failed |
| `sitk_only` | SimpleITK succeeded; PyDICOM pixel read failed |
| `metadata_ok_pixels_failed` | Metadata readable; neither tool decoded pixels |
| `empty_file` | File has 0 bytes — incomplete download |
| `proprietary_format` | Non-standard binary (e.g. Philips 3D/4D proprietary container) |
| `both_failed_metadata` | Neither tool could read even the metadata |

```python
def classify(probe, meta, pixel, sk):
    """Derive comparison_status and root_cause from actual probe/read results."""
    pydicom_meta_ok = meta.get("pydicom_meta_ok", False)
    pydicom_pixel_ok = pixel.get("pydicom_pixel_ok", False)
    s_ok = sk.get("sitk_ok", False)
    sig = probe.get("binary_signature", "")

    if pydicom_pixel_ok and s_ok:
        status = "both_ok"
    elif sig == "empty_file":
        status = "empty_file"
    elif sig == "philips_private_3d":
        status = "proprietary_format"
    elif pydicom_pixel_ok and not s_ok:
        status = "pydicom_only"
    elif s_ok and not pydicom_pixel_ok:
        status = "sitk_only"
    elif pydicom_meta_ok and not pydicom_pixel_ok and not s_ok:
        status = "metadata_ok_pixels_failed"
    else:
        status = "both_failed_metadata"

    if sig == "empty_file":
        cause = (
            "Empty file (0 bytes). Has a .dcm extension but no content. "
            "Likely an incomplete download from the source dataset. "
            "Verify dataset integrity and re-download the missing files."
        )
    elif sig == "philips_private_3d":
        cause = (
            "Philips proprietary 3D/4D format (magic bytes 377ABCAF, private tag 7A37,AFBC). "
            "Not standard DICOM — this is a Philips-specific container for 3D/4D ultrasound volumes. "
            "MicroDICOM has a dedicated decoder for this format. "
            "PyDICOM and SimpleITK cannot decode it without a Philips vendor plugin."
        )
    elif sig == "implicit_vr_candidate" and not pydicom_pixel_ok:
        cause = (
            "DICOM file without Part 10 preamble (Implicit VR Little Endian). "
            "MicroDICOM reads it via GDCM. Fix: pip install gdcm. "
            f"PyDICOM error: {meta.get('pydicom_meta_error', '')[:100]}. "
            f"SimpleITK error: {sk.get('sitk_error', '')[:80]}."
        )
    elif sig == "unknown_binary" and not pydicom_meta_ok:
        cause = (
            "No DICM prefix and no recognisable DICOM tags. "
            "File is likely corrupt or not DICOM. "
            f"PyDICOM error: {meta.get('pydicom_meta_error', '')[:100]}. "
            f"SimpleITK error: {sk.get('sitk_error', '')[:80]}."
        )
    elif not pydicom_pixel_ok and pydicom_meta_ok:
        ts = meta.get("TransferSyntaxName", meta.get("TransferSyntaxUID", "unknown"))
        pix_err = pixel.get("pydicom_pixel_error", "")
        sig_note = " (no DICM prefix, opened with force=True)" if sig == "unknown_binary" else ""
        cause = (
            f"Metadata readable{sig_note}. "
            f"Pixel data not decodable — transfer syntax: '{ts}'. "
            f"MicroDICOM decodes this format internally via GDCM. "
            f"Fix: pip install gdcm. "
            f"PyDICOM error: {pix_err[:120]}."
        )
    elif not pydicom_meta_ok:
        cause = f"PyDICOM could not read metadata. Error: {meta.get('pydicom_meta_error', 'unknown')[:120]}."
    elif pydicom_pixel_ok and not s_ok:
        cause = f"PyDICOM succeeded. SimpleITK failed: {sk.get('sitk_error', '')[:120]}."
    else:
        cause = "Both tools read the file successfully. No issues detected."

    return status, cause


rows = []
for f in candidate_files:
    key = str(f)
    probe = probe_results.get(key, {})
    meta  = meta_results.get(key, {})
    pixel = pixel_results.get(key, {})
    sk    = sitk_results.get(key, {})
    status, cause = classify(probe, meta, pixel, sk)
    try:
        rel = str(f.relative_to(BASE_DIR))
    except ValueError:
        rel = f.name
    rows.append({
        "file": rel,
        "comparison_status": status,
        "binary_signature": probe.get("binary_signature", ""),
        "manufacturer": meta.get("Manufacturer", ""),
        "manufacturer_model": meta.get("ManufacturerModelName", ""),
        "sop_class": meta.get("SOPClassName", meta.get("SOPClassUID", "")),
        "transfer_syntax": meta.get("TransferSyntaxName", ""),
        "photometric": meta.get("PhotometricInterpretation", ""),
        "n_frames": meta.get("NumberOfFrames", ""),
        "pydicom_pixel_ok": pixel.get("pydicom_pixel_ok", False),
        "pydicom_pixel_shape": pixel.get("pydicom_pixel_shape", ""),
        "pydicom_pixel_warnings": pixel.get("pydicom_pixel_warnings", ""),
        "sitk_ok": sk.get("sitk_ok", False),
        "sitk_size": sk.get("sitk_size", ""),
        "root_cause": cause,
    })

df = pd.DataFrame(rows)
from IPython.display import display
print(f"\nStatus summary ({len(df)} files):")
print(df["comparison_status"].value_counts().to_string())
print("\nSignature breakdown:")
print(df["binary_signature"].value_counts().to_string())
print("\nDetailed table:")
display(df[["file", "comparison_status", "binary_signature", "manufacturer", "sop_class", "transfer_syntax"]].to_html(index=False))
```

## Section 8 — Doppler / Colour-Flow Evidence

We detect colour-flow / Doppler candidates using only DICOM tags — no pixel heuristics.

```python
def doppler_evidence(meta: dict) -> tuple:
    """Returns (is_candidate: bool, evidence: list[str])."""
    evidence = []
    if str(meta.get("UltrasoundColorDataPresent", "")).strip() == "1":
        evidence.append("UltrasoundColorDataPresent=1 (explicit DICOM tag)")
    photometric = meta.get("PhotometricInterpretation", "").upper()
    try:
        spp = int(float(str(meta.get("SamplesPerPixel", "0"))))
    except (ValueError, TypeError):
        spp = 0
    colour_photometrics = {"PALETTE COLOR", "RGB", "YBR_FULL", "YBR_FULL_422"}
    if photometric in colour_photometrics and spp >= 1:
        evidence.append(f"PhotometricInterpretation={photometric} with SamplesPerPixel={spp}")
    doppler_keywords = {"doppler", "color flow", "colour flow", "cfm", "color map"}
    for tag in ["SeriesDescription", "StudyDescription", "ProtocolName", "ImageType"]:
        val = meta.get(tag, "").lower()
        for kw in doppler_keywords:
            if kw in val:
                evidence.append(f"{tag} contains '{kw}'")
                break
    return bool(evidence), evidence


doppler_rows = []
for f in candidate_files:
    key = str(f)
    meta = meta_results.get(key, {})
    is_candidate, evidence_list = doppler_evidence(meta)
    try:
        label = str(f.relative_to(BASE_DIR))
    except ValueError:
        label = f.name
    doppler_rows.append({
        "file": label,
        "doppler_candidate": is_candidate,
        "evidence": " | ".join(evidence_list) if evidence_list else "no validated evidence",
        "photometric": meta.get("PhotometricInterpretation", ""),
        "samples_per_pixel": meta.get("SamplesPerPixel", ""),
        "UltrasoundColorDataPresent": meta.get("UltrasoundColorDataPresent", ""),
    })

df_doppler = pd.DataFrame(doppler_rows)
candidates = df_doppler[df_doppler["doppler_candidate"]]
non_candidates = df_doppler[~df_doppler["doppler_candidate"]]

print(f"Confirmed Doppler/colour-flow candidates: {len(candidates)}")
for _, row in candidates.iterrows():
    print(f"  {row['file']}")
    print(f"  Evidence: {row['evidence']}\n")

print(f"Files with NO validated Doppler/colour evidence: {len(non_candidates)}")
for _, row in non_candidates.iterrows():
    print(f"  {row['file']}  ({row['photometric']})")
```

## Section 9 — Preliminary Channel-First Shape Mapping

We compute the expected tensor shape for each successfully decoded file as MONAI would receive it.

**Shape convention used here:**  
MONAI's `LoadImage` / `EnsureChannelFirst` pipeline expects `(C, [D,] H, W)`.  
We derive this from the PyDICOM pixel array shape (which is the ground truth):

| PyDICOM pixel_array shape | Interpretation | MONAI expected shape |
|---|---|---|
| `(H, W)` | Single-frame, 1 channel | `(1, H, W)` |
| `(H, W, C)` | Single-frame, C channels | `(C, H, W)` |
| `(F, H, W)` | Multi-frame, 1 channel | `(1, F, H, W)` |
| `(F, H, W, C)` | Multi-frame, C channels | `(C, F, H, W)` |

**Note:** No shape is reported if pixel decoding failed. This section does not create a torch.Tensor or run a MONAI transform.

```python
def monai_shape(shape_str: str) -> dict:
    """Derive channel-first shape candidate from PyDICOM pixel_array shape string."""
    result = {"raw_array_shape": shape_str, "monai_shape_CHW": "", "acquisition_type": "", "shape_mappable": False, "shape_note": ""}
    if not shape_str:
        result["shape_note"] = "No pixel data decoded."
        return result
    try:
        dims = [int(x) for x in shape_str.split("x")]
    except Exception:
        result["shape_note"] = f"Could not parse shape: {shape_str}"
        return result
    ndim = len(dims)
    if ndim == 2:
        H, W = dims
        result.update(monai_shape_CHW=f"(1, {H}, {W})", acquisition_type="2D single-frame, 1 channel", shape_mappable=True)
    elif ndim == 3:
        d0, d1, d2 = dims
        if d2 <= 4:
            result.update(monai_shape_CHW=f"({d2}, {d0}, {d1})", acquisition_type="2D single-frame, multi-channel (C last in array)", shape_mappable=True)
        else:
            result.update(monai_shape_CHW=f"(1, {d0}, {d1}, {d2})", acquisition_type="3D volume or multi-frame, 1 channel", shape_mappable=True)
        result["shape_note"] = "Validate C vs F interpretation against SamplesPerPixel tag."
    elif ndim == 4:
        F, H, W, C = dims
        result.update(monai_shape_CHW=f"({C}, {F}, {H}, {W})", acquisition_type="Multi-frame, multi-channel (C last in array)", shape_mappable=True)
    else:
        result["shape_note"] = f"Unexpected ndim={ndim}. Manual inspection required."
    return result


tensor_rows = []
for f in candidate_files:
    key = str(f)
    pixel = pixel_results.get(key, {})
    meta  = meta_results.get(key, {})
    try:
        label = str(f.relative_to(BASE_DIR))
    except ValueError:
        label = f.name
    shape_info = monai_shape(pixel.get("pydicom_pixel_shape", ""))
    spp_note = ""
    spp_raw = meta.get("SamplesPerPixel", "")
    if spp_raw and shape_info["shape_mappable"]:
        try:
            spp = int(float(spp_raw))
            dims = [int(x) for x in pixel.get("pydicom_pixel_shape", "").split("x")]
            if dims and dims[-1] == spp:
                spp_note = f"Channel count validated against SamplesPerPixel={spp}"
            elif spp == 1:
                spp_note = f"SamplesPerPixel={spp}; last array dim may be frames, not channels"
        except Exception:
            pass
    tensor_rows.append({
        "file": label,
        "raw_array_shape": shape_info["raw_array_shape"],
        "monai_shape_CHW": shape_info["monai_shape_CHW"],
        "acquisition_type": shape_info["acquisition_type"],
        "shape_mappable": shape_info["shape_mappable"],
        "spp_validation": spp_note,
    })

df_tensor = pd.DataFrame(tensor_rows)
ready = df_tensor[df_tensor["shape_mappable"]]
not_ready = df_tensor[~df_tensor["shape_mappable"]]

print(f"Files with determinable preliminary channel-first shape: {len(ready)}")
display(ready[["file", "raw_array_shape", "monai_shape_CHW", "acquisition_type", "spp_validation"]].to_html(index=False))

if not not_ready.empty:
    print(f"\nFiles where tensor shape could NOT be determined: {len(not_ready)}")
    for _, row in not_ready.iterrows():
        print(f"  {row['file']}")
```

```python
# ============================================================
# DICOM DATASET COLLECTION CONFIG
# ============================================================

from pathlib import Path
import pydicom


# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

BASE_DIR = Path(".").resolve()

DATASET = BASE_DIR / "Datasets"

WORKSPACE_DIR = BASE_DIR.parent


# ------------------------------------------------------------
# INPUT DATASET FOLDERS
# ------------------------------------------------------------

DATASET_ROOTS = [
    DATASET / "Aliza",
    DATASET / "AIIMS",
    DATASET / "Breast_Lesions_USG",
    DATASET / "MONAI",
    DATASET / "ReMIND",
    DATASET / "Heartcycle",
    DATASET / "UTA4",
    DATASET / "UTA7",
    DATASET / "UTA10",
    DATASET / "TCIA_IDC_Prostate_MRI_US_Biopsy",
    
]


# ------------------------------------------------------------
# OUTPUT CSV
# ------------------------------------------------------------

OUT_CSV = BASE_DIR / "interoperability_evaluation_us_dicom_tensor.csv"


# ============================================================
# STRICT DICOM DETECTION
# ============================================================

def is_dicom_file(path: Path) -> bool:
    """
    Strict DICOM validation.

    Prevents false positives from:
    - PNG
    - JPG
    - MP4
    - random binary files
    """

    try:

        # Standard DICOM preamble check
        with open(path, "rb") as f:
            f.seek(128)
            magic = f.read(4)

        if magic == b"DICM":
            return True

        # Secondary strict validation
        ds = pydicom.dcmread(
            str(path),
            stop_before_pixels=True,
            force=False
        )

        # Essential DICOM tags must exist
        required_tags = [
            "PatientID",
            "StudyInstanceUID",
            "SeriesInstanceUID",
            "SOPInstanceUID",
        ]

        return any(hasattr(ds, tag) for tag in required_tags)

    except Exception:
        return False


# ============================================================
# COLLECT DICOM FILES
# ============================================================

def collect_dicom_files(roots):

    found = []

    print("Collecting DICOM files...\n")

    for root in roots:

        if not root.exists():
            print(f"[NOT FOUND] {root}")
            continue

        print(f"Scanning: {root}")

        all_files = [f for f in root.rglob("*") if f.is_file()]

        before_count = len(found)

        for f in all_files:

            # Common DICOM extensions
            if f.suffix.lower() in [".dcm", ".dicom"]:
                if is_dicom_file(f):
                    found.append(f)
                continue

            # Extensionless anonymized DICOMs
            if f.suffix == "":
                if is_dicom_file(f):
                    found.append(f)

        added = len(found) - before_count

        print(f"  Found {added} DICOM files\n")

    return sorted(found)


# ============================================================
# RUN COLLECTION
# ============================================================

candidate_files = collect_dicom_files(DATASET_ROOTS)

print("=" * 60)
print(f"TOTAL DICOM FILES FOUND: {len(candidate_files)}")
print("=" * 60)
```

## Section 2 — Structural Probe

Before calling PyDICOM or SimpleITK we check the raw binary signature of each file.

**Note:**  
If a file lacks the DICOM Part 10 preamble (`DICM` at byte offset 128), it may still be a valid DICOM implicit-VR file — but if it also has no recoverable UIDs and no file-meta information, both libraries will fail.  
We record this **before** any library call to explain failures accurately.

```python
def structural_probe(path: Path) -> dict:
    """Read first 256 bytes and check DICOM structural markers."""
    result = {"dicm_prefix_present": False, "dicm_offset": -1, "file_meta_likely": False, "binary_signature": "unknown", "probe_error": ""}
    try:
        raw = path.read_bytes()[:256]
        if raw[128:132] == b"DICM":
            result["dicm_prefix_present"] = True
            result["dicm_offset"] = 128
            result["binary_signature"] = "dicom_part10"
        else:
            try:
                text = raw.decode("latin-1")
            except Exception:
                text = ""
            uid_like = any(c.isdigit() or c == "." for c in text[8:30]) and "." in text[8:30]
            result["binary_signature"] = "implicit_vr_candidate" if uid_like else "unknown_binary"
        result["file_meta_likely"] = b"\x02\x00" in raw[:136]
    except Exception as exc:
        result["probe_error"] = f"{type(exc).__name__}: {exc}"
    return result

print("Structural probe results:")
print(f"{'File':<55} {'DICM':>6} {'Signature'}")
print("-" * 90)
probe_results = {str(f): structural_probe(f) for f in candidate_files}
for f in candidate_files:
    pr = probe_results[str(f)]
    try:
        label = str(f.relative_to(BASE_DIR))
    except ValueError:
        label = f.name
    print(f"{label:<55} {str(pr['dicm_prefix_present']):>6}  {pr['binary_signature']}")
```

## Section 3 — PyDICOM: Metadata Read

We attempt to read DICOM metadata with PyDICOM.  
We try **without** `force=True` first, then **with** `force=True` if that fails.  


```python
import pandas as pd
import warnings
import traceback
import csv

TAGS_OF_INTEREST = [
    "Modality", "Manufacturer", "ManufacturerModelName",
    "SOPClassUID", "TransferSyntaxUID",
    "Rows", "Columns", "SamplesPerPixel",
    "PhotometricInterpretation", "NumberOfFrames", "UltrasoundColorDataPresent", "ImageType",
    "StudyDescription", "SeriesDescription", "ProtocolName",
    "VolumetricProperties", "FrameTime", "CineRate",
    "BitsAllocated", "BitsStored", "HighBit", "PixelRepresentation",
]

def classify_failure(error_msg, warnings_str):
    if not error_msg:
        return ("OK", "")
    if "InvalidDicomError" in error_msg:
        return ("Not DICOM", "File does not conform to DICOM Part 10 or is corrupted")
    if "FileNotFoundError" in error_msg:
        return ("File Not Found", "File missing or path error")
    if "ValueError" in error_msg and "DICM" in error_msg:
        return ("Missing DICM Prefix", "No DICM marker at offset 128 (not a Part 10 file)")
    if "UnsupportedTransferSyntax" in error_msg or "decoder not available" in error_msg:
        return ("Decoder Missing", "Compressed transfer syntax not supported by installed decoders")
    if "AttributeError" in error_msg and "file_meta" in error_msg:
        return ("Missing File Meta", "No file meta information (0002,xxxx) present")
    if "force=True" in error_msg:
        return ("Severely Corrupted", "Cannot recover even with force=True")
    if "PermissionError" in error_msg:
        return ("Permission Error", "No read access to file")
    if warnings_str:
        return ("Warning", warnings_str)
    return ("Other Error", error_msg)

def _get_tag(ds, keyword, default=""):
    try:
        val = getattr(ds, keyword)
        if hasattr(val, '__iter__') and not isinstance(val, str):
            return "\\".join(str(v) for v in val)
        return str(val) if val is not None else default
    except AttributeError:
        return default

def pydicom_read_metadata(path: Path) -> dict:
    result = {"pydicom_meta_ok": False, "pydicom_meta_forced": False, "pydicom_meta_error": "", "pydicom_warnings": ""}
    ds = None
    with warnings.catch_warnings(record=True) as w:
        warnings.simplefilter("always")
        try:
            ds = pydicom.dcmread(str(path), stop_before_pixels=True, force=False)
            result["pydicom_meta_ok"] = True
        except Exception as exc:
            result["pydicom_meta_error"] = f"{type(exc).__name__}: {exc}"
            try:
                ds = pydicom.dcmread(str(path), stop_before_pixels=True, force=True)
                result["pydicom_meta_ok"] = True
                result["pydicom_meta_forced"] = True
                result["pydicom_meta_error"] = ""
            except Exception as exc2:
                result["pydicom_meta_error"] = (
                    f"standard: {type(exc).__name__}: {exc} | "
                    f"force=True: {type(exc2).__name__}: {exc2}"
                )
        if w:
            result["pydicom_warnings"] = "; ".join(f"{x.category.__name__}: {x.message}" for x in w)
    if ds is not None:
        for tag in TAGS_OF_INTEREST:
            result[tag] = _get_tag(ds, tag)
        try:
            result["TransferSyntaxName"] = UID(ds.file_meta.TransferSyntaxUID).name \
                if hasattr(ds, "file_meta") and hasattr(ds.file_meta, "TransferSyntaxUID") else ""
        except Exception:
            result["TransferSyntaxName"] = ""
        try:
            result["SOPClassName"] = UID(ds.SOPClassUID).name
        except Exception:
            result["SOPClassName"] = ""
    return result

print("Reading metadata with PyDICOM and logging failures...")
meta_results = {}
failure_log = []
for f in candidate_files:
    r = pydicom_read_metadata(f)
    meta_results[str(f)] = r
    try:
        label = str(f.relative_to(BASE_DIR))
    except ValueError:
        label = f.name
    status = "OK (forced)" if r["pydicom_meta_forced"] else ("OK" if r["pydicom_meta_ok"] else "FAILED")
    error_hint = f" -> {r['pydicom_meta_error'][:80]}" if not r["pydicom_meta_ok"] else ""
    fail_type, fail_root = classify_failure(r["pydicom_meta_error"], r["pydicom_warnings"])
    if status.startswith("OK"):
        print(f"  {label:<55} {status}")
    else:
        print(f"  {label:<55} {status} [{fail_type}] {fail_root}")
        failure_log.append({
            "file": str(f),
            "status": status,
            "failure_type": fail_type,
            "root_cause": fail_root,
            "error": r["pydicom_meta_error"],
            "warnings": r["pydicom_warnings"],
        })

print(f"\nTotal failures: {len(failure_log)}")

fail_report_path = BASE_DIR / f"failure_report_{len(candidate_files)}files_{len(failure_log)}failures_{pd.Timestamp.now().strftime('%Y%m%d')}.csv"
with open(fail_report_path, "w", newline='', encoding="utf-8") as csvfile:
    fieldnames = ["file", "status", "failure_type", "root_cause", "error", "warnings"]
    writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
    writer.writeheader()
    for row in failure_log:
        writer.writerow(row)
print(f"Failure report exported to: {fail_report_path}")

problem_dir = BASE_DIR / f"problematic_files_{pd.Timestamp.now().strftime('%Y%m%d')}"
problem_dir.mkdir(exist_ok=True)
for fail in failure_log:
    failtype_dir = problem_dir / fail["failure_type"].replace(" ", "_")
    failtype_dir.mkdir(exist_ok=True)
    src = Path(fail["file"])
    dst = failtype_dir / src.name
    try:
        if not dst.exists():
            shutil.copy2(src, dst)
    except Exception as e:
        print(f"[COPY ERROR] {src} -> {dst}: {e}")

print(f"Problematic files copied to: {problem_dir}")
print("\nDone.")
```

## Section 4 — PyDICOM: Pixel Data Read

We attempt to decode the actual pixel array.  
This is a **separate** step from metadata because metadata can succeed while pixel decoding fails 

When decoding fails, the cell shows the **exact error** and the **transfer syntax** that caused it so you know which optional decoder is needed.

```python
def pydicom_read_pixels(path: Path) -> dict:
    result = {
        "pydicom_pixel_ok": False,
        "pydicom_pixel_error": "",
        "pydicom_pixel_shape": "",
        "pydicom_pixel_dtype": "",
        "pydicom_pixel_warnings": "",
    }
    with warnings.catch_warnings(record=True) as w:
        warnings.simplefilter("always")
        try:
            ds = pydicom.dcmread(str(path), force=False)
            arr = ds.pixel_array
            result["pydicom_pixel_ok"] = True
            result["pydicom_pixel_shape"] = "x".join(str(d) for d in arr.shape)
            result["pydicom_pixel_dtype"] = str(arr.dtype)
        except Exception as exc:
            result["pydicom_pixel_error"] = f"{type(exc).__name__}: {exc}"
        if w:
            result["pydicom_pixel_warnings"] = "; ".join(
                f"{x.category.__name__}: {x.message}" for x in w
            )
    return result

print("Reading pixel data with PyDICOM...")
pixel_results = {}
for f in candidate_files:
    # Only attempt pixel read if metadata was readable
    if not meta_results[str(f)]["pydicom_meta_ok"]:
        pixel_results[str(f)] = {
            "pydicom_pixel_ok": False,
            "pydicom_pixel_error": "SKIPPED — metadata read failed",
            "pydicom_pixel_shape": "",
            "pydicom_pixel_dtype": "",
            "pydicom_pixel_warnings": "",
        }
        continue
    r = pydicom_read_pixels(f)
    pixel_results[str(f)] = r
    try:
        label = str(f.relative_to(BASE_DIR))
    except ValueError:
        label = f.name
    ts = meta_results[str(f)].get("TransferSyntaxName", "")
    if r["pydicom_pixel_ok"]:
        warn_flag = " [warnings present]" if r["pydicom_pixel_warnings"] else ""
        print(f"  {label:<55} OK  shape={r['pydicom_pixel_shape']}  dtype={r['pydicom_pixel_dtype']}{warn_flag}")
    else:
        print(f"  {label:<55} FAILED  [{ts}]")
        print(f"    Reason: {r['pydicom_pixel_error'][:120]}")

print("\nDone.")
```

## Section 5 — SimpleITK Read

A file can be readable by PyDICOM but not by SimpleITK depending on the transfer syntax and how each library handles it.  

```python
def sitk_read(path: Path) -> dict:
    result = {
        "sitk_ok": False,
        "sitk_error": "",
        "sitk_dimension": "",
        "sitk_size": "",
    }
    if not SITK_OK:
        result["sitk_error"] = "SimpleITK not installed"
        return result
    try:
        reader = sitk.ImageFileReader()
        reader.SetFileName(str(path))
        reader.ReadImageInformation()
        img = reader.Execute()
        result["sitk_ok"] = True
        result["sitk_dimension"] = str(img.GetDimension())
        result["sitk_size"] = "x".join(str(s) for s in img.GetSize())
    except Exception as exc:
        result["sitk_error"] = f"{type(exc).__name__}: {exc}"
    return result

print("Reading files with SimpleITK...")
sitk_results = {}
for f in candidate_files:
    r = sitk_read(f)
    sitk_results[str(f)] = r
    try:
        label = str(f.relative_to(BASE_DIR))
    except ValueError:
        label = f.name
    if r["sitk_ok"]:
        print(f"  {label:<55} OK  dim={r['sitk_dimension']}  size={r['sitk_size']}")
    else:
        print(f"  {label:<55} FAILED")
        print(f"    Reason: {r['sitk_error'][:120]}")

print("\nDone.")
```

## Section 6 — Consistency Classification

We combine the results from Sections 2–5 into a single per-file classification.

| `comparison_status` | Meaning |
|---|---|
| `both_ok` | Both PyDICOM and SimpleITK read the file successfully |
| `pydicom_only` | PyDICOM read pixels; SimpleITK failed |
| `sitk_only` | SimpleITK succeeded; PyDICOM pixel read failed |
| `metadata_ok_pixels_failed` | Both read metadata; neither decoded pixels |
| `both_failed_metadata` | Neither tool could read even the metadata |

```python
def classify(probe, meta, pixel, sk):
    """Derive comparison_status and root_cause from actual probe/read results."""
    pydicom_meta_ok = meta.get("pydicom_meta_ok", False)
    pydicom_pixel_ok = pixel.get("pydicom_pixel_ok", False)
    sitk_ok = sk.get("sitk_ok", False)

    if pydicom_pixel_ok and sitk_ok:
        status = "both_ok"
    elif pydicom_pixel_ok and not sitk_ok:
        status = "pydicom_only"
    elif sitk_ok and not pydicom_pixel_ok:
        status = "sitk_only"
    elif pydicom_meta_ok and not pydicom_pixel_ok and not sitk_ok:
        status = "metadata_ok_pixels_failed"
    else:
        status = "both_failed_metadata"

    if probe["binary_signature"] == "unknown_binary":
        cause = (
            "File lacks DICOM Part 10 DICM prefix, file-meta block, and recoverable "
            "core tags. This is either a corrupt file or a non-DICOM binary. "
            f"PyDICOM error (force=True): {meta.get('pydicom_meta_error', '')[:120]}. "
            f"SimpleITK error: {sk.get('sitk_error', '')[:80]}."
        )
    elif not pydicom_pixel_ok and pydicom_meta_ok:
        ts = meta.get("TransferSyntaxName", meta.get("TransferSyntaxUID", "unknown"))
        pix_err = pixel.get("pydicom_pixel_error", "")
        cause = (
            f"Metadata readable. Pixel decoding failed. "
            f"Transfer syntax: '{ts}'. "
            f"PyDICOM pixel error: {pix_err[:120]}. "
            "Install 'pylibjpeg' or 'gdcm' to decode compressed data."
        )
    elif not pydicom_meta_ok:
        cause = (
            f"PyDICOM could not read metadata. "
            f"Error: {meta.get('pydicom_meta_error', 'unknown')[:120]}."
        )
    elif pydicom_pixel_ok and not sitk_ok:
        cause = (
            f"PyDICOM succeeded. SimpleITK failed: {sk.get('sitk_error', '')[:120]}."
        )
    else:
        cause = "Both tools read the file successfully. No issues detected."

    return status, cause

rows = []
for f in candidate_files:
    key = str(f)
    probe = probe_results.get(key, {})
    meta  = meta_results.get(key, {})
    pixel = pixel_results.get(key, {})
    sk    = sitk_results.get(key, {})
    status, cause = classify(probe, meta, pixel, sk)
    try:
        rel = str(f.relative_to(BASE_DIR))
    except ValueError:
        rel = f.name
    rows.append({
        "file": rel,
        "comparison_status": status,
        "dicm_prefix": probe.get("dicm_prefix_present", False),
        "binary_signature": probe.get("binary_signature", ""),
        "manufacturer": meta.get("Manufacturer", ""),
        "sop_class": meta.get("SOPClassName", meta.get("SOPClassUID", "")),
        "transfer_syntax": meta.get("TransferSyntaxName", ""),
        "rows": meta.get("Rows", ""),
        "columns": meta.get("Columns", ""),
        "samples_per_pixel": meta.get("SamplesPerPixel", ""),
        "photometric": meta.get("PhotometricInterpretation", ""),
        "n_frames": meta.get("NumberOfFrames", ""),
        "pydicom_pixel_ok": pixel.get("pydicom_pixel_ok", False),
        "pydicom_pixel_shape": pixel.get("pydicom_pixel_shape", ""),
        "pydicom_pixel_warnings": pixel.get("pydicom_pixel_warnings", ""),
        "sitk_ok": sk.get("sitk_ok", False),
        "sitk_size": sk.get("sitk_size", ""),
        "root_cause": cause,
    })

df = pd.DataFrame(rows)
from IPython.display import display, HTML
print(f"\nStatus summary ({len(df)} files):")
print(df["comparison_status"].value_counts().to_string())
print("\nDetailed table:")
display(df[["file", "comparison_status", "manufacturer", "sop_class", "transfer_syntax"]].to_html(index=False))
```

## Section 8 — Doppler / Colour-Flow Evidence


```python
def doppler_evidence(meta: dict) -> tuple:
    """
    Returns (is_candidate: bool, evidence: list[str]).
    Only uses facts from the actual DICOM metadata — no assumptions.
    """
    evidence = []

    # Rule 1: explicit DICOM tag
    ucdp = meta.get("UltrasoundColorDataPresent", "")
    if str(ucdp).strip() == "1":
        evidence.append("UltrasoundColorDataPresent=1 (explicit DICOM tag)")

    # Rule 2: photometric interpretation + samples per pixel
    photometric = meta.get("PhotometricInterpretation", "").upper()
    spp_raw = meta.get("SamplesPerPixel", "")
    try:
        spp = int(float(str(spp_raw)))
    except (ValueError, TypeError):
        spp = 0
    colour_photometrics = {"PALETTE COLOR", "RGB", "YBR_FULL", "YBR_FULL_422"}
    if photometric in colour_photometrics and spp >= 1:
        evidence.append(f"PhotometricInterpretation={photometric} with SamplesPerPixel={spp}")

    # Rule 3: keyword evidence in text tags
    doppler_keywords = {"doppler", "color flow", "colour flow", "cfm", "color map"}
    for tag in ["SeriesDescription", "StudyDescription", "ProtocolName", "ImageType"]:
        val = meta.get(tag, "").lower()
        for kw in doppler_keywords:
            if kw in val:
                evidence.append(f"{tag} contains '{kw}'")
                break

    return bool(evidence), evidence

print("Doppler / colour-flow candidate analysis:\n")
doppler_rows = []
for f in candidate_files:
    key = str(f)
    meta = meta_results.get(key, {})
    is_candidate, evidence_list = doppler_evidence(meta)
    try:
        label = str(f.relative_to(BASE_DIR))
    except ValueError:
        label = f.name
    doppler_rows.append({
        "file": label,
        "doppler_candidate": is_candidate,
        "evidence": " | ".join(evidence_list) if evidence_list else "no validated evidence",
        "photometric": meta.get("PhotometricInterpretation", ""),
        "samples_per_pixel": meta.get("SamplesPerPixel", ""),
        "UltrasoundColorDataPresent": meta.get("UltrasoundColorDataPresent", ""),
    })

df_doppler = pd.DataFrame(doppler_rows)

candidates = df_doppler[df_doppler["doppler_candidate"]]
non_candidates = df_doppler[~df_doppler["doppler_candidate"]]

print(f"Confirmed Doppler/colour-flow candidates: {len(candidates)}")
for _, row in candidates.iterrows():
    print(f"  {row['file']}")
    print(f"  Evidence: {row['evidence']}")
    print()

print(f"Files with NO validated Doppler/colour evidence: {len(non_candidates)}")
for _, row in non_candidates.iterrows():
    print(f"  {row['file']}  ({row['photometric']})")
```

## Section 9 — Preliminary Channel-First Shape Mapping

We compute the expected tensor shape for each successfully decoded file as MONAI would receive it.

**Shape convention used here:**  
MONAI's `LoadImage` / `EnsureChannelFirst` pipeline expects `(C, [D,] H, W)`.  
We derive this from the PyDICOM pixel array shape (which is the ground truth):

| PyDICOM pixel_array shape | Interpretation | MONAI expected shape |
|---|---|---|
| `(H, W)` | Single-frame, 1 channel | `(1, H, W)` |
| `(H, W, C)` | Single-frame, C channels | `(C, H, W)` |
| `(F, H, W)` | Multi-frame, 1 channel | `(1, F, H, W)` or `(F, 1, H, W)` |
| `(F, H, W, C)` | Multi-frame, C channels | `(C, F, H, W)` |

**Note:** The shape shown here is derived from the actual decoded array.  
No shape is reported if pixel decoding failed. This section does not create a torch.Tensor or run a MONAI transform.

```python
def monai_shape(shape_str: str, n_frames_meta: str) -> dict:
    """
    Derive shape-mappable array from PyDICOM pixel_array shape string.
    shape_str: e.g. '292x600x800x3' from pydicom_pixel_shape
    """
    result = {
        "raw_array_shape": shape_str,
        "ndim": "",
        "monai_shape_CHW": "",
        "acquisition_type": "",
        "shape_mappable": False,
        "shape_note": "",
    }
    if not shape_str:
        result["shape_note"] = "No pixel data decoded — shape cannot be determined."
        return result
    try:
        dims = [int(x) for x in shape_str.split("x")]
    except Exception:
        result["shape_note"] = f"Could not parse shape string: {shape_str}"
        return result

    ndim = len(dims)
    result["ndim"] = ndim

    if ndim == 2:
        H, W = dims
        result["monai_shape_CHW"] = f"(1, {H}, {W})"
        result["acquisition_type"] = "2D single-frame, 1 channel"
        result["shape_mappable"] = True
    elif ndim == 3:
        # Could be (H, W, C) or (F, H, W)
        d0, d1, d2 = dims
        # Heuristic: last dim <= 4 is likely channels; validated against SamplesPerPixel later
        if d2 <= 4:
            result["monai_shape_CHW"] = f"({d2}, {d0}, {d1})"
            result["acquisition_type"] = "2D single-frame, multi-channel (C last in array)"
        else:
            result["monai_shape_CHW"] = f"(1, {d0}, {d1}, {d2})"
            result["acquisition_type"] = "3D volume or multi-frame, 1 channel"
        result["shape_mappable"] = True
        result["shape_note"] = (
            "3D array: last-dim <= 4 interpreted as channels; otherwise as spatial/temporal. "
            "Validate against SamplesPerPixel tag."
        )
    elif ndim == 4:
        F, H, W, C = dims
        result["monai_shape_CHW"] = f"({C}, {F}, {H}, {W})"
        result["acquisition_type"] = "Multi-frame, multi-channel (C last in array)"
        result["shape_mappable"] = True
    else:
        result["shape_note"] = f"Unexpected number of dimensions: {ndim}. Manual inspection required."

    return result

print("preliminary channel-first shape analysis:\n")
tensor_rows = []
for f in candidate_files:
    key = str(f)
    pixel = pixel_results.get(key, {})
    meta  = meta_results.get(key, {})
    try:
        label = str(f.relative_to(BASE_DIR))
    except ValueError:
        label = f.name

    shape_info = monai_shape(
        pixel.get("pydicom_pixel_shape", ""),
        meta.get("NumberOfFrames", "")
    )
    # Cross-validate channel count with SamplesPerPixel
    spp_raw = meta.get("SamplesPerPixel", "")
    spp_note = ""
    if spp_raw and shape_info["shape_mappable"]:
        try:
            spp = int(float(spp_raw))
            dims = [int(x) for x in pixel.get("pydicom_pixel_shape", "").split("x")]
            if len(dims) >= 2 and dims[-1] == spp:
                spp_note = f"Channel count validated against SamplesPerPixel={spp}"
            elif spp == 1 and len(dims) in (2, 3):
                spp_note = f"SamplesPerPixel={spp}; last array dim may be frames, not channels"
        except Exception:
            pass

    tensor_rows.append({
        "file": label,
        "raw_array_shape": shape_info["raw_array_shape"],
        "monai_shape_CHW": shape_info["monai_shape_CHW"],
        "acquisition_type": shape_info["acquisition_type"],
        "shape_mappable": shape_info["shape_mappable"],
        "spp_validation": spp_note,
        "shape_note": shape_info["shape_note"],
    })

df_tensor = pd.DataFrame(tensor_rows)

ready = df_tensor[df_tensor["shape_mappable"]]
not_ready = df_tensor[~df_tensor["shape_mappable"]]

print(f"Files with determinable preliminary channel-first shape: {len(ready)}")
display(ready[["file", "raw_array_shape", "monai_shape_CHW", "acquisition_type", "spp_validation"]].to_html(index=False))

if not not_ready.empty:
    print(f"\nFiles where tensor shape could NOT be determined: {len(not_ready)}")
    for _, row in not_ready.iterrows():
        print(f"  {row['file']}: {row['shape_note']}")
```

## Section 10 — Final Summary Table

# Validated MONAI Data Contract

This section validates five distinct readiness levels instead of treating them as equivalent:

1. **Pixel decodable:** `pydicom.pixel_array` can decode the stored pixels.
2. **Shape mappable:** the decoded array can be mapped to a channel-first layout.
3. **Tensor created:** a real `torch.Tensor` is created with channel and frame checks.
4. **MONAI validated:** the tensor passes through a MONAI dictionary transform as a `MetaTensor`, with metadata preserved.
5. **Semantically validated:** DICOM evidence is sufficient to classify the object as `single_frame_2d`, `temporal_cine`, or `spatial_volume_3d`.

`undetermined_multiframe` is used whenever metadata cannot distinguish the meaning of the frame axis, or when spatial and temporal evidence coexist and require a more detailed dimension-index reconstruction. Non-ultrasound objects are excluded before pixel/tensor validation.


```python
from pathlib import Path
import gc
import importlib.util

import numpy as np
import pandas as pd
import pydicom
import torch
from pydicom.uid import UID
from IPython.display import display

if importlib.util.find_spec("monai") is None:
    raise ImportError("MONAI is required for the validated data-contract section.")

import monai
from monai.data import MetaTensor
from monai.transforms import Compose, EnsureTyped

try:
    from pydicom.pixels import apply_color_lut
except ImportError:  # pydicom < 3
    from pydicom.pixel_data_handlers.util import apply_color_lut

print(f"torch: {torch.__version__}")
print(f"MONAI: {monai.__version__}")
print(f"pydicom: {pydicom.__version__}")

US_SOP_CLASS_UIDS = {
    "1.2.840.10008.5.1.4.1.1.6.1",  # Ultrasound Image Storage
    "1.2.840.10008.5.1.4.1.1.3.1",  # Ultrasound Multi-frame Image Storage
    "1.2.840.10008.5.1.4.1.1.6.2",  # Enhanced US Volume Storage
}
ENHANCED_US_VOLUME_UID = "1.2.840.10008.5.1.4.1.1.6.2"
TEMPORAL_POINTER_TAGS = {"(0018,1063)", "(0018,1065)"}  # FrameTime / FrameTimeVector

# Set either limit to an integer for a quick development run. None validates all files.
VALIDATION_MAX_FILES = None
VALIDATION_MAX_US_FILES = None


def _text(value, default=""):
    if value is None:
        return default
    if isinstance(value, (list, tuple)):
        return "\\".join(str(item) for item in value)
    return str(value)


def _positive_int(value, default=1):
    try:
        return max(1, int(value))
    except (TypeError, ValueError):
        return default


def _existing_candidate_files():
    existing = globals().get("candidate_files", [])
    return sorted({Path(item).resolve() for item in existing if Path(item).is_file()})


def _fallback_candidate_files():
    base = Path(".").resolve()
    roots = [
        base / "Datasets",
        base / "DICOM AND DIRECT DATA" / "ALIZA",
        base / "DICOM AND DIRECT DATA" / "remind",
    ]
    files = []
    for root in roots:
        if root.exists():
            files.extend(path.resolve() for path in root.rglob("*.dcm") if path.is_file())
            files.extend(path.resolve() for path in root.rglob("*.dicom") if path.is_file())
    return sorted(set(files))


def validation_candidate_files():
    files = _existing_candidate_files() or _fallback_candidate_files()
    if VALIDATION_MAX_FILES is not None:
        files = files[:VALIDATION_MAX_FILES]
    if not files:
        raise FileNotFoundError(
            "No DICOM candidates found. Run dataset discovery first or place data under "
            "./Datasets or ./DICOM AND DIRECT DATA/{ALIZA,remind}."
        )
    return files


def ultrasound_identity(ds):
    modality = _text(getattr(ds, "Modality", "")).strip().upper()
    sop_uid = _text(getattr(ds, "SOPClassUID", "")).strip()
    evidence = []
    if modality == "US":
        evidence.append("Modality=US")
    if sop_uid in US_SOP_CLASS_UIDS:
        evidence.append(f"ultrasound SOP Class={UID(sop_uid).name}")
    return bool(evidence), evidence


def _shared_item(ds):
    sequence = getattr(ds, "SharedFunctionalGroupsSequence", None)
    return sequence[0] if sequence else None


def _first_per_frame_item(ds):
    sequence = getattr(ds, "PerFrameFunctionalGroupsSequence", None)
    return sequence[0] if sequence else None


def _nested_volumetric_properties(ds):
    values = []
    top_level = _text(getattr(ds, "VolumetricProperties", "")).strip().upper()
    if top_level:
        values.append(top_level)
    shared = _shared_item(ds)
    description = getattr(shared, "USImageDescriptionSequence", None) if shared else None
    if description:
        nested = _text(getattr(description[0], "VolumetricProperties", "")).strip().upper()
        if nested:
            values.append(nested)
    return values


def classify_ultrasound_semantics(ds):
    n_frames = _positive_int(getattr(ds, "NumberOfFrames", 1))
    if n_frames <= 1:
        return {
            "semantic_type": "single_frame_2d",
            "classification_evidence": "NumberOfFrames absent or equal to 1",
            "semantically_validated": True,
            "uncertainty_reason": "",
        }

    sop_uid = _text(getattr(ds, "SOPClassUID", "")).strip()
    temporal_evidence = []
    spatial_evidence = []

    for keyword in ("FrameTime", "FrameTimeVector", "CineRate", "RecommendedDisplayFrameRate"):
        if getattr(ds, keyword, None) not in (None, "", []):
            temporal_evidence.append(f"{keyword} present")

    frame_pointer = _text(getattr(ds, "FrameIncrementPointer", ""))
    if any(tag in frame_pointer for tag in TEMPORAL_POINTER_TAGS):
        temporal_evidence.append(f"FrameIncrementPointer={frame_pointer}")

    shared = _shared_item(ds)
    first_frame = _first_per_frame_item(ds)
    if (shared and hasattr(shared, "TemporalPositionSequence")) or (
        first_frame and hasattr(first_frame, "TemporalPositionSequence")
    ):
        temporal_evidence.append("TemporalPositionSequence present")

    volumetric_values = _nested_volumetric_properties(ds)
    if "VOLUME" in volumetric_values:
        spatial_evidence.append("VolumetricProperties=VOLUME")
    if sop_uid == ENHANCED_US_VOLUME_UID:
        spatial_evidence.append("Enhanced US Volume Storage SOP Class")
    if (shared and hasattr(shared, "PlaneOrientationVolumeSequence")) or (
        first_frame and hasattr(first_frame, "PlanePositionVolumeSequence")
    ):
        spatial_evidence.append("volume plane functional groups present")
    image_type = _text(getattr(ds, "ImageType", "")).upper()
    if "VOLUME" in image_type:
        spatial_evidence.append("ImageType contains VOLUME")

    if temporal_evidence and spatial_evidence:
        return {
            "semantic_type": "undetermined_multiframe",
            "classification_evidence": " | ".join(spatial_evidence + temporal_evidence),
            "semantically_validated": False,
            "uncertainty_reason": (
                "Both spatial and temporal evidence are present; the flattened frame axis must be "
                "reconstructed from DimensionIndexSequence before assigning T and Z axes."
            ),
        }
    if temporal_evidence:
        return {
            "semantic_type": "temporal_cine",
            "classification_evidence": " | ".join(temporal_evidence),
            "semantically_validated": True,
            "uncertainty_reason": "",
        }
    if spatial_evidence:
        return {
            "semantic_type": "spatial_volume_3d",
            "classification_evidence": " | ".join(spatial_evidence),
            "semantically_validated": True,
            "uncertainty_reason": "",
        }
    return {
        "semantic_type": "undetermined_multiframe",
        "classification_evidence": f"NumberOfFrames={n_frames}, without explicit temporal or spatial evidence",
        "semantically_validated": False,
        "uncertainty_reason": "DICOM metadata does not establish whether the frame axis is time or space.",
    }


def decoded_pixels(ds):
    array = ds.pixel_array
    processing = "pydicom pixel_array"
    if _text(getattr(ds, "PhotometricInterpretation", "")).upper() == "PALETTE COLOR":
        array = apply_color_lut(array, ds)
        processing += " + DICOM palette LUT to RGB"
    return np.asarray(array), processing


def channel_first_tensor(array, ds):
    n_frames = _positive_int(getattr(ds, "NumberOfFrames", 1))
    samples_per_pixel = _positive_int(getattr(ds, "SamplesPerPixel", 1))
    source_shape = tuple(int(size) for size in array.shape)

    if n_frames > 1:
        if array.ndim == 3 and samples_per_pixel == 1:
            channel_first = array[np.newaxis, ...]
        elif array.ndim == 4 and array.shape[-1] in (3, 4):
            channel_first = np.moveaxis(array, -1, 0)
        else:
            raise ValueError(
                f"Unsupported multiframe layout {source_shape} for SamplesPerPixel={samples_per_pixel}."
            )
        frame_validation_ok = int(channel_first.shape[1]) == n_frames
    else:
        if array.ndim == 2:
            channel_first = array[np.newaxis, ...]
        elif array.ndim == 3 and array.shape[-1] in (3, 4):
            channel_first = np.moveaxis(array, -1, 0)
        else:
            raise ValueError(
                f"Unsupported single-frame layout {source_shape} for SamplesPerPixel={samples_per_pixel}."
            )
        frame_validation_ok = True

    channel_first = np.ascontiguousarray(channel_first)
    tensor = torch.as_tensor(channel_first)
    photometric = _text(getattr(ds, "PhotometricInterpretation", "")).upper()
    expected_channels = 3 if photometric == "PALETTE COLOR" else samples_per_pixel
    channel_validation_ok = int(tensor.shape[0]) == expected_channels

    return tensor, channel_validation_ok, frame_validation_ok


def metadata_for_tensor(path, ds, semantic):
    sop_uid = _text(getattr(ds, "SOPClassUID", ""))
    try:
        sop_name = UID(sop_uid).name if sop_uid else ""
    except Exception:
        sop_name = ""
    return {
        "filename_or_obj": str(path),
        "modality": _text(getattr(ds, "Modality", "")),
        "sop_class_uid": sop_uid,
        "sop_class_name": sop_name,
        "study_instance_uid": _text(getattr(ds, "StudyInstanceUID", "")),
        "series_instance_uid": _text(getattr(ds, "SeriesInstanceUID", "")),
        "sop_instance_uid": _text(getattr(ds, "SOPInstanceUID", "")),
        "manufacturer": _text(getattr(ds, "Manufacturer", "")),
        "manufacturer_model": _text(getattr(ds, "ManufacturerModelName", "")),
        "number_of_frames": _positive_int(getattr(ds, "NumberOfFrames", 1)),
        "samples_per_pixel": _positive_int(getattr(ds, "SamplesPerPixel", 1)),
        "photometric_interpretation": _text(getattr(ds, "PhotometricInterpretation", "")),
        "frame_time": _text(getattr(ds, "FrameTime", "")),
        "cine_rate": _text(getattr(ds, "CineRate", "")),
        "volumetric_properties": "\\".join(_nested_volumetric_properties(ds)),
        "semantic_type": semantic["semantic_type"],
        "classification_evidence": semantic["classification_evidence"],
    }


MONAI_CONTRACT = Compose([
    EnsureTyped(keys="image", data_type="tensor", track_meta=True),
])


def load_monai_ultrasound_sample(path):
    """Return one validated MONAI sample on demand without retaining the full corpus in memory."""
    path = Path(path).resolve()
    ds_meta = pydicom.dcmread(str(path), stop_before_pixels=True, force=False)
    is_us, identity_evidence = ultrasound_identity(ds_meta)
    if not is_us:
        raise ValueError(f"Not identified as ultrasound by DICOM metadata: {path}")

    semantic = classify_ultrasound_semantics(ds_meta)
    ds = pydicom.dcmread(str(path), force=False)
    array, pixel_processing = decoded_pixels(ds)
    tensor, channel_ok, frame_ok = channel_first_tensor(array, ds)
    metadata = metadata_for_tensor(path, ds, semantic)
    sample = MONAI_CONTRACT({"image": MetaTensor(tensor, meta=metadata)})

    diagnostics = {
        "identity_evidence": " | ".join(identity_evidence),
        "semantic_type": semantic["semantic_type"],
        "semantically_validated": semantic["semantically_validated"],
        "uncertainty_reason": semantic["uncertainty_reason"],
        "raw_shape": tuple(int(size) for size in array.shape),
        "tensor_shape": tuple(int(size) for size in sample["image"].shape),
        "tensor_dtype": str(sample["image"].dtype),
        "pixel_processing": pixel_processing,
        "channel_validation_ok": bool(channel_ok),
        "frame_validation_ok": bool(frame_ok),
    }
    return sample, diagnostics
```

```python
def validate_dicom_for_monai(path):
    row = {
        "file": str(path),
        "modality": "",
        "sop_class_uid": "",
        "sop_class_name": "",
        "is_ultrasound": False,
        "semantic_type": "unclassified",
        "classification_evidence": "",
        "pixel_decodable": False,
        "shape_mappable": False,
        "tensor_created": False,
        "raw_shape": "",
        "tensor_shape": "",
        "tensor_dtype": "",
        "pixel_processing": "",
        "channel_validation_ok": False,
        "frame_validation_ok": False,
        "metadata_attached": False,
        "monai_validation_ok": False,
        "semantically_validated": False,
        "uncertainty_reason": "",
        "error": "",
    }

    try:
        ds_meta = pydicom.dcmread(str(path), stop_before_pixels=True, force=False)
    except Exception as exc:
        row["error"] = f"Metadata read failed: {type(exc).__name__}: {exc}"
        row["uncertainty_reason"] = "Object cannot be classified without readable DICOM metadata."
        return row

    modality = _text(getattr(ds_meta, "Modality", "")).strip().upper()
    sop_uid = _text(getattr(ds_meta, "SOPClassUID", "")).strip()
    try:
        sop_name = UID(sop_uid).name if sop_uid else ""
    except Exception:
        sop_name = ""
    row.update(modality=modality, sop_class_uid=sop_uid, sop_class_name=sop_name)

    is_us, identity_evidence = ultrasound_identity(ds_meta)
    row["is_ultrasound"] = is_us
    if not is_us:
        row["semantic_type"] = "not_ultrasound"
        row["classification_evidence"] = f"Modality={modality or 'missing'} | SOP Class={sop_name or sop_uid or 'missing'}"
        return row

    semantic = classify_ultrasound_semantics(ds_meta)
    row.update(
        semantic_type=semantic["semantic_type"],
        classification_evidence=" | ".join(identity_evidence + [semantic["classification_evidence"]]),
        semantically_validated=semantic["semantically_validated"],
        uncertainty_reason=semantic["uncertainty_reason"],
    )

    try:
        ds = pydicom.dcmread(str(path), force=False)
        array, pixel_processing = decoded_pixels(ds)
        row["pixel_decodable"] = True
        row["raw_shape"] = str(tuple(int(size) for size in array.shape))
        row["pixel_processing"] = pixel_processing

        tensor, channel_ok, frame_ok = channel_first_tensor(array, ds)
        row["shape_mappable"] = True
        row["tensor_created"] = isinstance(tensor, torch.Tensor)
        row["tensor_shape"] = str(tuple(int(size) for size in tensor.shape))
        row["tensor_dtype"] = str(tensor.dtype)
        row["channel_validation_ok"] = bool(channel_ok)
        row["frame_validation_ok"] = bool(frame_ok)

        metadata = metadata_for_tensor(path, ds, semantic)
        sample = {"image": MetaTensor(tensor, meta=metadata)}
        validated = MONAI_CONTRACT(sample)
        output = validated["image"]

        required_meta = {"filename_or_obj", "modality", "sop_class_uid", "semantic_type"}
        row["metadata_attached"] = isinstance(output, MetaTensor) and required_meta.issubset(output.meta)
        row["monai_validation_ok"] = bool(
            isinstance(output, MetaTensor)
            and row["metadata_attached"]
            and tuple(output.shape) == tuple(tensor.shape)
            and row["channel_validation_ok"]
            and row["frame_validation_ok"]
        )
    except Exception as exc:
        row["error"] = f"Pixel/tensor/MONAI validation failed: {type(exc).__name__}: {exc}"
    finally:
        for variable in ("array", "tensor", "sample", "validated", "output", "ds"):
            if variable in locals():
                del locals()[variable]
        gc.collect()

    return row


validation_files = validation_candidate_files()
validation_rows = []
ultrasound_seen = 0

for file_path in validation_files:
    result = validate_dicom_for_monai(file_path)
    if result["is_ultrasound"]:
        if VALIDATION_MAX_US_FILES is not None and ultrasound_seen >= VALIDATION_MAX_US_FILES:
            continue
        ultrasound_seen += 1
    validation_rows.append(result)

validated_contract_df = pd.DataFrame(validation_rows)
print(f"Validated candidate files: {len(validated_contract_df)}")
print(f"Ultrasound objects identified by DICOM metadata: {int(validated_contract_df['is_ultrasound'].sum())}")
```

```python
def _rate(column, denominator):
    if denominator == 0:
        return "n/a"
    return f"{100.0 * int(column.sum()) / denominator:.1f}%"

us_contract_df = validated_contract_df[validated_contract_df["is_ultrasound"]].copy()
non_us_contract_df = validated_contract_df[~validated_contract_df["is_ultrasound"]].copy()
us_count = len(us_contract_df)

readiness_summary = pd.DataFrame([
    {"readiness_level": "pixel_decodable", "count": int(us_contract_df["pixel_decodable"].sum()), "rate_among_US": _rate(us_contract_df["pixel_decodable"], us_count)},
    {"readiness_level": "shape_mappable", "count": int(us_contract_df["shape_mappable"].sum()), "rate_among_US": _rate(us_contract_df["shape_mappable"], us_count)},
    {"readiness_level": "tensor_created", "count": int(us_contract_df["tensor_created"].sum()), "rate_among_US": _rate(us_contract_df["tensor_created"], us_count)},
    {"readiness_level": "metadata_attached", "count": int(us_contract_df["metadata_attached"].sum()), "rate_among_US": _rate(us_contract_df["metadata_attached"], us_count)},
    {"readiness_level": "monai_validation_ok", "count": int(us_contract_df["monai_validation_ok"].sum()), "rate_among_US": _rate(us_contract_df["monai_validation_ok"], us_count)},
    {"readiness_level": "semantically_validated", "count": int(us_contract_df["semantically_validated"].sum()), "rate_among_US": _rate(us_contract_df["semantically_validated"], us_count)},
])

print("Modality counts (classification occurs before tensor creation):")
display(validated_contract_df["modality"].replace("", "<missing>").value_counts(dropna=False).rename_axis("modality").reset_index(name="count"))

print("Ultrasound semantic classes:")
display(us_contract_df["semantic_type"].value_counts(dropna=False).rename_axis("semantic_type").reset_index(name="count"))

print("Readiness levels for DICOM objects identified as ultrasound:")
display(readiness_summary)

final_columns = [
    "file", "modality", "sop_class_name", "is_ultrasound", "semantic_type",
    "classification_evidence", "raw_shape", "tensor_shape", "tensor_dtype",
    "pixel_decodable", "shape_mappable", "tensor_created", "channel_validation_ok",
    "frame_validation_ok", "metadata_attached", "monai_validation_ok",
    "semantically_validated", "uncertainty_reason", "error",
]
print("Per-ultrasound-object validation table:")
display(us_contract_df[final_columns])

# Keep one small, real MetaTensor accessible as an example of the output contract.
example_candidates = us_contract_df[us_contract_df["semantic_type"] == "single_frame_2d"]["file"].tolist()
validated_example_sample = None
validated_example_diagnostics = None
if example_candidates:
    example_path = min(example_candidates, key=lambda item: Path(item).stat().st_size)
    validated_example_sample, validated_example_diagnostics = load_monai_ultrasound_sample(example_path)
    print("Accessible MetaTensor example:")
    print("  file:", example_path)
    print("  shape:", tuple(validated_example_sample["image"].shape))
    print("  metadata keys:", sorted(validated_example_sample["image"].meta.keys()))

print("Methodological conclusion:")
print(
    "Pixel decoding, shape mapping, tensor creation, MONAI validation, and semantic validation "
    "are reported separately. A file is not described as fully MONAI-ready only because its "
    "pixel array has a mappable shape."
)
```

## Interpretation of the validated contract

- **Non-ultrasound modalities are excluded first.** This prevents MRI, segmentation, mammography, or unreadable objects from inflating ultrasound readiness results.
- **A real tensor is created.** `torch.as_tensor()` receives the decoded array after channel-first reordering; palette-color objects are expanded with the DICOM color lookup table.
- **MONAI is exercised directly.** Each ultrasound sample passes through `EnsureTyped` and remains a `MetaTensor` with selected non-PHI technical metadata attached.
- **Frame meaning is evidence-based.** Temporal tags support `temporal_cine`; volumetric tags and functional groups support `spatial_volume_3d`. Mixed or insufficient evidence produces `undetermined_multiframe`.
- **Readiness is layered.** The final report does not use a single “100% MONAI-ready” claim. It reports decoding, mapping, tensor creation, metadata retention, MONAI validation, and semantic validation independently.
